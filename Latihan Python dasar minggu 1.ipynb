```python
def hitung_return(harga_kemarin, harga_hari_ini):
  return (harga_hari_ini - harga_kemarin) / harga_kemarin * 100

print(hitung_return(2630, 2645))
```

```python

harga_5_hari = [2643, 2651, 2693, 2700, 2630]
total = 0
for harga in harga_5_hari:
  total += harga

rata_rata = total / len(harga_5_hari)
print("Rata-rata harga:", rata_rata)
```

```python
for i in range(1, len(harga_5_hari)):
  harga_hari_ini = harga_5_hari[i]
  harga_kemarin = harga_5_hari[i - 1]

  if harga_hari_ini > harga_kemarin:
    print(f"Hari ke-{i + 1}: NAIK")
  elif harga_hari_ini < harga_kemarin:
    print(f"Hari ke-{i + 1}: TURUN")
  else:
    print(f"Hari ke-{i + 1}: TETAP")
```

```python
def cari_hari_terbaik(daftar_harga):
  harga_terbesar = daftar_harga[0]

  for harga in daftar_harga:
    if harga > harga_terbesar:
      harga_terbesar = harga

  return harga_terbesar
hasil = cari_hari_terbaik(harga_5_hari)
print("Harga harian terbesar:", hasil)
```
